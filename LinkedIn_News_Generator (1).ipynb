```python
import feedparser
from datetime import datetime

# Get ESG-related news
def fetch_esg_news(max_articles=5):
    rss_url = "https://news.google.com/rss/search?q=ESG+investing+OR+sustainable+finance&hl=en"
    feed = feedparser.parse(rss_url)
    return feed.entries[:max_articles]

# Format prompts for ChatGPT
def create_chatgpt_prompts():
    articles = fetch_esg_news()
    prompts = []
    for article in articles:
        title = article.title
        url = article.link
        prompt = (
            f"Write a short, professional LinkedIn post (2–3 sentences) summarizing this ESG-related headline:\n\n"
            f"Title: {title}\nLink: {url}\n\n"
            f"Keep it insightful and suitable for an ESG investing audience. Include a call-to-action like "
            f"'Check how your portfolio compares with ESG trends at MacroBloom.com.'"
        )
        prompts.append(prompt)
    return prompts

# Print prompts to copy into ChatGPT
if __name__ == "__main__":
    prompts = create_chatgpt_prompts()
    today = datetime.today().strftime('%Y-%m-%d')
    print(f"\n🔁 ESG ChatGPT Prompts for {today}:\n")
    for i, p in enumerate(prompts, 1):
        print(f"--- Prompt #{i} ---\n{p}\n")
```
